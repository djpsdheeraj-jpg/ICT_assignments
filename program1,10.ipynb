```python
# Smart Library Circulation System (Without Functions)

MAX_BOOKS_PER_STUDENT = 3
ALLOWED_DAYS = 14
FINE_PER_DAY = 2

books = {}
students = {}

while True:
    print("\n--- Smart Library Circulation System ---")
    print("1. Add Book")
    print("2. Register Student")
    print("3. Issue Book")
    print("4. Return Book")
    print("5. Search Books")
    print("6. Overdue Report")
    print("7. Exit")

    choice = input("Enter your choice: ").strip()

    # 1. Add Book
    if choice == "1":
        book_id = input("Enter Book ID: ").strip()
        if book_id in books:
            print("Book ID already exists.")
            continue

        title = input("Enter Title: ").strip()
        author = input("Enter Author: ").strip()

        try:
            total = int(input("Enter Total Copies: "))
            if total <= 0:
                raise ValueError
        except ValueError:
            print("Invalid number of copies.")
            continue

        books[book_id] = {
            "title": title,
            "author": author,
            "total": total,
            "available": total
        }
        print("Book added successfully.")

    # 2. Register Student
    elif choice == "2":
        student_id = input("Enter Student ID: ").strip()
        if student_id in students:
            print("Student already registered.")
            continue

        name = input("Enter Student Name: ").strip()
        students[student_id] = {
            "name": name,
            "issued_books": {}
        }
        print("Student registered successfully.")

    # 3. Issue Book
    elif choice == "3":
        student_id = input("Enter Student ID: ").strip()
        book_id = input("Enter Book ID: ").strip()

        if student_id not in students:
            print("Student not found.")
            continue
        if book_id not in books:
            print("Book not found.")
            continue

        if books[book_id]["available"] <= 0:
            print("No copies available.")
            continue

        if len(students[student_id]["issued_books"]) >= MAX_BOOKS_PER_STUDENT:
            print("Student has reached maximum book limit.")
            continue

        if book_id in students[student_id]["issued_books"]:
            print("Same book already issued to this student.")
            continue

        students[student_id]["issued_books"][book_id] = 0
        books[book_id]["available"] -= 1
        print("Book issued successfully.")

    # 4. Return Book
    elif choice == "4":
        student_id = input("Enter Student ID: ").strip()
        book_id = input("Enter Book ID: ").strip()

        if student_id not in students:
            print("Student not found.")
            continue

        if book_id not in students[student_id]["issued_books"]:
            print("This book was not issued to the student.")
            continue

        try:
            days = int(input("Enter number of days borrowed: "))
            if days < 0:
                raise ValueError
        except ValueError:
            print("Invalid number of days.")
            continue

        overdue_days = max(0, days - ALLOWED_DAYS)
        fine = overdue_days * FINE_PER_DAY

        books[book_id]["available"] += 1
        del students[student_id]["issued_books"][book_id]

        print("Book returned successfully.")
        print("Overdue Fine:", fine)

    # 5. Search Books (Partial Matching)
    elif choice == "5":
        keyword = input("Enter keyword to search: ").lower()
        found = False

        for book_id, data in books.items():
            if keyword in data["title"].lower() or keyword in data["author"].lower():
                print(f"{book_id} | {data['title']} | {data['author']} | Available: {data['available']}")
                found = True

        if not found:
            print("No matching books found.")

    # 6. Overdue Report
    elif choice == "6":
        print("\n--- Overdue Report ---")
        report_found = False

        for sid, student in students.items():
            for bid, days in student["issued_books"].items():
                if days > ALLOWED_DAYS:
                    overdue = days - ALLOWED_DAYS
                    fine = overdue * FINE_PER_DAY
                    print(f"Student: {student['name']} | Book: {books[bid]['title']} | Overdue Days: {overdue} | Fine: {fine}")
                    report_found = True

        if not report_found:
            print("No overdue books.")

    # 7. Exit
    elif choice == "7":
        print("Exiting Smart Library System.")
        break

    else:
        print("Invalid choice. Try again.")
```

```python
# Sample log data (can be read from a file later)
logs = [
    "2026-01-01 10:15:30 INFO User login Alice",
    "2026-01-01 10:16:10 ERROR Disk failure",
    "2026-01-01 10:17:05 WARNING Low memory",
    "2026-01-01 10:18:00 INFO User logout Alice",
    "2026-01-01 10:19:45 ERROR Network timeout"
]

severity_count = {}
error_types = {}
user_activity = {}

def analyze_logs():
    for log in logs:
        parts = log.split()
        severity = parts[2]

        # Count severity
        severity_count[severity] = severity_count.get(severity, 0) + 1

        # Count errors
        if severity == "ERROR":
            error = " ".join(parts[3:])
            error_types[error] = error_types.get(error, 0) + 1

        # User activity
        if "User" in log:
            user = parts[-1]
            user_activity[user] = user_activity.get(user, 0) + 1

def filter_by_severity(level):
    print("\nFiltered Logs:")
    for log in logs:
        if level in log:
            print(log)

analyze_logs()

while True:
    print("\n--- Log File Analyzer ---")
    print("1. Show severity count")
    print("2. Show error types")
    print("3. Show user activity")
    print("4. Filter logs by severity")
    print("5. Exit")

    choice = input("Enter choice: ")

    if choice == "1":
        print(severity_count)

    elif choice == "2":
        print(error_types)

    elif choice == "3":
        print(user_activity)

    elif choice == "4":
        level = input("Enter severity (INFO/ERROR/WARNING): ")
        filter_by_severity(level)

    elif choice == "5":
        break

    else:
        print("Invalid choice")
```
